# Метрика

Перед решением любой задачи важно понимать, как будет оцениваться ваше решение. В данном случае мы используем стандартную для задачи классификации метрику ROC-AUC. Ее можно вычислить, используя только предсказанные вероятности и истинные классы без конкретного порога классификации + она раотает даже если классы в данных сильно несбалансированны (примеров одного класса в десятки раз больше примеров длугого). Именно поэтому она очень удобна для соревнований.

Посчитать ее легко:


```python
from sklearn.metrics import roc_auc_score

y_true = [
    0,
    1,
    1,
    0,
    1
]

y_predictions = [
    0.1,
    0.9,
    0.4,
    0.6,
    0.61
]

roc_auc_score(y_true, y_predictions)
```

## Применение градиентного бустинга (2 балла)

Если вы хотите получить баллы за точный ответ, то стоит попробовать градиентный бустинг. Часто градиентный бустинг с параметрами по умолчанию даст вам 80% результата за 0% усилий.

Мы будем использовать catboost, поэтому нам не надо кодировать категориальные признаки. Catboost сделает это сам (в .fit() надо передать cat_features=cat_cols). А численные признаки нормировать для моделей, основанных на деревьях не нужно.

1) Разделите выборку на train/valid. Протестируйте catboost cо стандартными параметрами.

2) Протестируйте разные занчения параметров количества деревьев и learning_rate'а и выберите лучшую по метрике ROC-AUC комбинацию.

(Дополнительно) Есть некоторые сложности с тем, чтобы использовать CatBoostClassifier вместе с GridSearchCV, поэтому мы не просим использовать кроссвалидацию. Но можете попробовать)

<font color="#16A085">Давайте потестим катбуст, правда надо будет подготовить данные, так как он работает с категориальными параметрами </font>

```python
data = pd.read_csv('./train.csv')
data_test = pd.read_csv('./test.csv')

data['TotalSpent'] = pd.to_numeric(data['TotalSpent'], errors='coerce')
data_test['TotalSpent'] = pd.to_numeric(data_test['TotalSpent'], errors='coerce')
data['TotalSpent'] = data['TotalSpent'].fillna(0)
data_test['TotalSpent'] = data_test['TotalSpent'].fillna(0)

cols_to_drop = ['Sex', 'HasPhoneService', 'HasMultiplePhoneNumbers',
                'HasOnlineTV', 'HasMovieSubscription', 'TotalSpent']

data = data.drop(columns=cols_to_drop)
data_test = data_test.drop(columns=cols_to_drop)

data_train_cat, data_valid_cat = train_test_split(data, test_size=0.3, random_state=92)
```

```python
cat_cols = [
    'HasPartner', 'HasChild', 'HasInternetService',
    'HasOnlineSecurityService', 'HasOnlineBackup',
    'HasDeviceProtection', 'HasTechSupportAccess',
    'HasContractPhone', 'IsBillingPaperless', 'PaymentMethod'
]

X_train_cat = data_train_cat.drop(columns=[target_col])
y_train_cat = data_train_cat[target_col]

X_valid_cat = data_valid_cat.drop(columns=[target_col])
y_valid_cat = data_valid_cat[target_col]
```

```python
!pip install catboost
```

```python
from catboost import CatBoostClassifier
from sklearn.metrics import roc_auc_score

model_catboost = CatBoostClassifier(
    random_seed=42,
    verbose=100
)

model_catboost.fit(
    X_train_cat, y_train_cat,
    cat_features=cat_cols,
    eval_set=(X_valid_cat, y_valid_cat),
    early_stopping_rounds=50,
    verbose=100
)

y_pred_proba = model_catboost.predict_proba(X_valid_cat)[:, 1]
print(f"ROC-AUC (default): {roc_auc_score(y_valid_cat, y_pred_proba):.4f}")
```

<font color="#16A085">Попробуем подобрать лучшие параметры </font>

```python
from itertools import product

n_estimators_list = [100, 300, 500, 1000]
learning_rate_list = [0.001, 0.01, 0.05, 0.07, 0.98, 0.1, 0.12, 0.3]

results = []

for n_est, lr in product(n_estimators_list, learning_rate_list):
    model = CatBoostClassifier(
        iterations=n_est,
        learning_rate=lr,
        random_seed=42,
        verbose=0
    )
    model.fit(X_train_cat, y_train_cat, cat_features=cat_cols, eval_set=(X_valid_cat, y_valid_cat))

    auc = roc_auc_score(y_valid_cat, model.predict_proba(X_valid_cat)[:, 1])
    results.append({'iterations': n_est, 'learning_rate': lr, 'roc_auc': auc})
    print(f"iterations={n_est}, lr={lr:.3f} → ROC-AUC={auc:.4f}")

results_df = pd.DataFrame(results)
best = results_df.loc[results_df['roc_auc'].idxmax()]
best_iterations = int(best['iterations'])
best_lr = best['learning_rate']
print(f"\nЛучшая комбинация:{best}")
```

```python
best_model = CatBoostClassifier(
    iterations = best_iterations,
    learning_rate = best_lr,
    cat_features=cat_cols,
    random_seed=42,
    verbose = 100,
    use_best_model=True,
)

best_model.fit(
    X_train_cat, y_train_cat,
    eval_set=(X_valid_cat, y_valid_cat),
    verbose=100
)

y_pred_proba = best_model.predict_proba(X_valid_cat)[:, 1]
final_auc = roc_auc_score(y_valid_cat, y_pred_proba)
print(f"Final ROC-AUC: {final_auc:.4f}")
```

<font color="#16A085">Значение метрики стало лучше </font>

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(
    y_valid_cat,
    y_pred_proba
)

plt.figure()
plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1])
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```
